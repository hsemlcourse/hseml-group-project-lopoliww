Baseline — DummyRegressor и Ridge без сложного feature engineering. Основная метрика: RMSE по log1p(learners_count).

```python
from pathlib import Path
import sys

# Если ноутбук открыт из папки notebooks, поднимаемся на уровень выше.
PROJECT_ROOT = Path.cwd()
if PROJECT_ROOT.name == "notebooks":
    PROJECT_ROOT = PROJECT_ROOT.parent

# Добавляем корень проекта в Python path.
if str(PROJECT_ROOT) not in sys.path:
    sys.path.insert(0, str(PROJECT_ROOT))

print("PROJECT_ROOT:", PROJECT_ROOT)
print("src exists:", (PROJECT_ROOT / "src").exists())
```

```python
from pathlib import Path
import pandas as pd

from src.modeling import load_dataset, split_dataset, build_preprocessor, evaluate
from sklearn.dummy import DummyRegressor
from sklearn.linear_model import Ridge
from sklearn.pipeline import Pipeline

DATA = Path('../data/processed/stepik_course_steps_processed.csv')
df = load_dataset(DATA)
data = split_dataset(df)
preprocessor = build_preprocessor(data.x_train)
```

```python
experiments = {
    'dummy_median': DummyRegressor(strategy='median'),
    'ridge_alpha_1': Ridge(alpha=1.0, random_state=42),
}
rows = []
for name, model in experiments.items():
    pipe = Pipeline([('preprocessor', preprocessor), ('model', model)])
    pipe.fit(data.x_train, data.y_train)
    rows.append({'model': name, **evaluate(pipe, data.x_val, data.y_val)})
pd.DataFrame(rows).sort_values('rmse_log')
```
